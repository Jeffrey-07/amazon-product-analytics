#Data Cleaning Steps

```python
df["discounted_price"]=(df["discounted_price"].str.replace("₹",'',regex=False).str.replace(",",'',regex=False).astype(float))
```

```python
df["actual_price"]=df["actual_price"].str.replace("₹",'',regex=False
                                                  ).str.replace(",",'',regex=False).astype(float)
```

```python
df.head()
```

```python
df["rating"]=pd.to_numeric(df["rating"],errors="coerce")
df["rating_count"]=pd.to_numeric(df["rating_count"],errors="coerce")
df["discount_percentage_numeric"] = pd.to_numeric(
    df["discount_percentage"].astype(str).str.replace("%", "", regex=False),
    errors="coerce"
)
```

```python
df["rating"]=df["rating"].fillna(df["rating"].median())
df["rating_count"]=df["rating_count"].fillna(df["rating_count"].median())
```

```python
df.isnull().sum()
```

#Preprocessing steps

```python
df['Main_category']=df['category'].str.split('|').str[0]
df['Sub_category']=df['category'].str.split('|').str[-1]
```

```python
df['money_saved']=df["actual_price"]-df["discounted_price"]
cat_count=df['Main_category'].value_counts()
print(cat_count)
```

#KPI'S

```python
tot_sales = len(df)
tot_cat = df["Main_category"].nunique()
avg_rating = df["rating"].mean()
avg_discount=df["discount_percentage_numeric"].mean()
rating_count = df["rating_count"].sum()
top_cat_product = df.groupby('Main_category')['money_saved'].sum().idxmax()
total_money_saved = df['money_saved'].sum()
avg_dis_percentage = df.groupby('Main_category')['discount_percentage_numeric'].mean().round(1)
print("Total Sales:",tot_sales)
print("Total Categories:",tot_cat)
print("Average Rating:",avg_rating)
print("Average Discount:",avg_discount)
print("Total Rating Count:",rating_count)
print("Top Category Product:",top_cat_product)
print("Total Money Saved:",total_money_saved)
print("Average Discount Percentage:",avg_dis_percentage)
```

```python
data_context=f"""
Total sales: {tot_sales}
total categories: {tot_cat}
average discount: {round(avg_discount, 1)}%
average rating: {round(avg_rating, 1)} out of 5
total money saved: {round(total_money_saved, 0)}
top category product: {top_cat_product}
discount percentage: {avg_dis_percentage}
"""
print(data_context)
```

```python
import os,getpass
os.environ["GROQ_API_KEY"]=getpass.getpass("paste your api key")
```

```python
from groq import Groq
client = Groq(api_key=os.getenv("GROQ_API_KEY"))    
```

```python
def ask_data(question):
    prompt = f"""
You are a data assistant for an Amazon dataset.
Answer the user's question only using the data provided in the context.

DATA:
{data_context}

QUESTION:
{question}

ANSWER:
"""

    try:
        resp = client.chat.completions.create(
            model="openai/gpt-oss-20b",
            messages=[
                {"role": "system", "content": prompt}
            ]
        )
        return resp.choices[0].message.content

    except Exception as e:
        return f"Error: {e}"
```

```python
print(ask_data("which category discounts the most?"))
print()
```

```python
print(ask_data("which category has most products?"))
print()
```
